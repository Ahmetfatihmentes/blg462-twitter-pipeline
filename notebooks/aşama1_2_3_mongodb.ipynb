```python
import pandas as pd

df = pd.read_csv('Tweets.csv', encoding='latin-1')
print("Temizleme öncesi kayıt sayısı:", len(df))
print("Boyut:", df.shape)
```

```python
import re

def temizle(metin):
    metin = re.sub(r'http\S+', '', str(metin))      # URL sil
    metin = re.sub(r'@\w+', '', metin)               # mention sil
    metin = re.sub(r'#\w+', '', metin)               # hashtag sil
    metin = re.sub(r'[^a-zA-Z\s]', '', metin)        # özel karakter sil
    return metin.strip()

print("Temizleme ÖNCESİ örnek:", df['text'][0])
df['text'] = df['text'].apply(temizle)
print("Temizleme SONRASI örnek:", df['text'][0])
```

```python
print("--- Null Değer Sayıları ---")
print(df.isnull().sum())
```

```python
# Kritik sütunlardaki null satırları sil
df = df.dropna(subset=['text', 'airline_sentiment', 'airline'])

# Diğer sütunları doldur
df['negativereason'] = df['negativereason'].fillna('unknown')
df['negativereason_confidence'] = df['negativereason_confidence'].fillna(0)
df['airline_sentiment_gold'] = df['airline_sentiment_gold'].fillna('unknown')
df['negativereason_gold'] = df['negativereason_gold'].fillna('unknown')
df['tweet_coord'] = df['tweet_coord'].fillna('unknown')
df['tweet_location'] = df['tweet_location'].fillna('unknown')
df['user_timezone'] = df['user_timezone'].fillna('unknown')

print("Null giderme sonrası kalan null sayısı:", df.isnull().sum().sum())
```

```python
print("Duplicate kayıt sayısı:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicate silme sonrası kayıt sayısı:", len(df))
```

```python
print("Dönüşüm ÖNCESİ tweet_created tipi:", df['tweet_created'].dtype)
df['tweet_created'] = pd.to_datetime(df['tweet_created'])
print("Dönüşüm SONRASI tweet_created tipi:", df['tweet_created'].dtype)
```

```python
print("--- Temizleme Özeti ---")
print("Temizleme öncesi kayıt sayısı: 14640")
print("Temizleme sonrası kayıt sayısı:", len(df))
print("Silinen kayıt sayısı:", 14640 - len(df))
```

```python
# Tamlık
tamlik_once = 71.78  # ham verideki doluluk oranı (yaklaşık)
tamlik_sonra = (df.notnull().sum().sum() / df.size) * 100

# Tutarlılık
tutarlilik = (df['tweet_created'].notna().sum() / len(df)) * 100

# Doğruluk
url_kalan = df['text'].str.contains(r'http', regex=True).sum()
dogruluk = ((len(df) - url_kalan) / len(df)) * 100

print("--- Veri Kalitesi Boyutları ---")
print(f"Tamlık (Completeness)   : Önce %{tamlik_once:.2f}  →  Sonra %{tamlik_sonra:.2f}")
print(f"Tutarlılık (Consistency): %{tutarlilik:.2f}  (tweet_created datetime formatında)")
print(f"Doğruluk (Accuracy)     : %{dogruluk:.2f}  (URL içermeyen temiz kayıt oranı)")

print("\n--- Temizlenmiş Veri Örnekleri (5 Kayıt) ---")
print(df[['tweet_id', 'text', 'tweet_created']].head(5).to_string(index=False))
```

```python
df.to_csv('tweets_temiz.csv', index=False)
print("tweets_temiz.csv kaydedildi!")
```

```python
# Havayoluna göre grupla, tweet sayısını hesapla, büyükten küçüğe sırala
pipeline = [
    {'$group': {'_id': '$airline', 'tweet_sayisi': {'$sum': 1}}},
    {'$sort': {'tweet_sayisi': -1}}
]
print("--- Sorgu 4: Havayoluna Göre Tweet Sayısı ---")
for doc in collection.aggregate(pipeline):
    print(doc)
```

```python
# Duygu durumlarını grupla, Türkçe alan adıyla çıkar
pipeline = [
    {'$group': {'_id': '$airline_sentiment', 'toplam': {'$sum': 1}}},
    {'$project': {'_id': 0, 'Duygu': '$_id', 'Toplam_Tweet': '$toplam'}}
]
print("--- Sorgu 5: Duygu Dağılımı ---")
for doc in collection.aggregate(pipeline):
    print(doc)
```

```python
# Önce negatif tweetleri filtrele, sonra havayoluna göre ortalama güven skoru hesapla
pipeline = [
    {'$match': {'airline_sentiment': 'negative'}},
    {'$group': {'_id': '$airline', 'ort_guven': {'$avg': '$airline_sentiment_confidence'}}},
    {'$sort': {'ort_guven': -1}}
]
print("--- Sorgu 6: Negatif Tweetlerde Ortalama Güven Skoru ---")
for doc in collection.aggregate(pipeline):
    print(doc)
```

```python
import time

# İndeks öncesi sorgu süresi
t1 = time.time()
collection.find_one({'airline': 'Delta'})
sure_once = round((time.time() - t1) * 1000, 2)
print(f"İndeks ÖNCESİ süre: {sure_once} ms")

# airline sütununa indeks oluştur
collection.create_index('airline')
print("İndeks oluşturuldu: airline")

# İndeks sonrası sorgu süresi
t2 = time.time()
collection.find_one({'airline': 'Delta'})
sure_sonra = round((time.time() - t2) * 1000, 2)
print(f"İndeks SONRASI süre: {sure_sonra} ms")
```

```python
# Delta'ya ait bir kaydı incelendi olarak işaretle
result = collection.update_one(
    {'airline': 'Delta'},
    {'$set': {'incelendi': True, 'not': 'Manuel inceleme yapıldı'}}
)
print("--- Sorgu 8: updateOne ---")
print("Eşleşen:", result.matched_count, "| Güncellenen:", result.modified_count)
```

```python
# Tüm neutral tweetlere reviewed alanı ekle
result = collection.update_many(
    {'airline_sentiment': 'neutral'},
    {'$set': {'reviewed': True}}
)
print("--- Sorgu 9: updateMany ---")
print("Eşleşen:", result.matched_count, "| Güncellenen:", result.modified_count)
```

```python
# Güven skoru 0.5 altındaki düşük güvenilirlikli kayıtları sil
result = collection.delete_many(
    {'airline_sentiment_confidence': {'$lt': 0.5}}
)
print("--- Sorgu 10: deleteMany ---")
print("Silinen kayıt sayısı:", result.deleted_count)
print("Kalan kayıt sayısı:", collection.count_documents({}))
```
